```python
import pandas as pd
```

```python
df = pd.read_csv("../dataset-documentation/cleaned_data/arrival_depart_cleaned_data/cleaned_2022-01_data.csv")
```

```python
def get_mean_interval(df):
    df["actual_datetime"] = pd.to_datetime(df["actual_datetime"])
    df['time_interval'] = df.groupby(['service_date', 'route_id', 'direction_id', 'stop_id'])['actual_datetime'].diff().dt.total_seconds()

    ret = df.groupby('stop_id', as_index=False).agg(mean_headway=("time_interval", "mean"), median_headway=("time_interval", "median"))

    return ret
```

```python
t = get_mean_interval(df)
```

```python
df.groupby("stop_id", as_index=False)["scheduled_headway"].nunique()
```

```python
df["headway_latency"] = df["time_interval"] - df["scheduled_headway"]
```

```python
df.groupby("stop_id", as_index=False)["headway_latency"].median().sort_values("headway_latency", ascending=False).to_csv("./test.csv", index=False)
```
